```python
import os
os.environ["VTK_DEFAULT_OPENGL_WINDOW"] = "vtkOSOpenGLRenderWindow"

from paraview.simple import *
import paraview.servermanager as sm

# --- CONNECT ONCE ---
if not sm.ActiveConnection:
    print("🔌 Connecting to built-in ParaView session...")
    Connect()
else:
    print("✅ ParaView session already active")

print("ParaView version:", GetParaViewVersion())
```

```python
def reset_pipeline():
    # Delete all sources
    for _, src in list(GetSources().items()):
        Delete(src)

    # Delete all views
    for v in list(GetViews()):
        Delete(v)

    print("♻️ Pipeline reset (server preserved)")
```

```python
from IPython.display import Image, display

reset_pipeline()

sphere = Sphere()
view = CreateView("RenderView")

rep = Show(sphere, view)
view.CameraPosition = [0, 0, 3]
view.CameraFocalPoint = [0, 0, 0]

Render()

fname = "/tmp/level1_local.png"
SaveScreenshot(fname, view, ImageResolution=[600, 400])
display(Image(filename=fname))

print("✅ Test complete: Local rendering")
```
